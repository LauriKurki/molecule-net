```python
import os
import tqdm
import tarfile
```

```python
def count_members(tarname):
    with tarfile.open(tarname) as t:
        return len(t.getmembers())
```

```python
path_to_tar = os.path.join("/l/data/molnet/SIN-STM-FDBM")
tar_files = [
    os.path.join(path_to_tar, f)
    for f in os.listdir(path_to_tar)
    if f.endswith(".tar")
]

count = {
    "train": 0,
    "val": 0,
    "test": 0,
}

for i, tar_file in enumerate(tqdm.tqdm(tar_files)):
    c = count_members(tar_file)
    if "train" in tar_file:
        count["train"] += c
    elif "val" in tar_file:
        count["val"] += c
    elif "test" in tar_file:
        count["test"] += c
    else:
        raise ValueError(f"Unknown split: {tar_file}")
    
    if i == 100:
        break
    
print(count)
```

```python
for k, v in count.items():
    print(f"{k}: {v/16}")
```
